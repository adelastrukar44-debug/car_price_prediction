```python
print("🔍🚗Car with the Highest Price.")
print(df.loc[df['price_usd'].idxmax()])
print("-"*40)
print("🔍🚗Car with the lowest Price.")
print(df.loc[df['price_usd'].idxmin()])
```

The highest price in the dataset is **$235,235**. After investigating the corresponding record, I found that it belongs to a **2018 Bentley Continental GT** with **29,000 km** mileage and a **6000 cm³** petrol engine.

The value appears to be plausible for this type of luxury vehicle, so it should not be automatically treated as an incorrect value. However, it represents an extreme value compared to most cars in the dataset and may influence the regression model.

### 🔎 Minimum Price Investigation

🔍🚗 Further Investigation -------

* Count total anomaly values.
* Display anomaly rows.

```python
print("Total number of rows with mileage_km in (9999999.0).")
print((df['mileage_km'] == 9999999.0).sum())
print("Total number of rows with mileage_km in (0.0)")
print((df['mileage_km'] == 0).sum())
```

```python
print(df[df['mileage_km'] == 9999999.0])
print("-"*40)
print(df[df['mileage_km'] == 0])
```

🔍🚗 Further Investigation – Extreme Mileage

After investigating all 22 vehicles with a mileage of 9,999,999 km, the value appears across different brands, models, years, prices, and vehicle conditions.
Since such a mileage value is unrealistic and occurs repeatedly across unrelated vehicles, it is most likely a placeholder or data entry anomaly rather than the actual mileage.
Therefore, these values should be treated as invalid mileage values during data preprocessing.

🔍🚗 Further Investigation – Zero Mileage

There are 190 vehicles with a mileage of 0 km. These values occur across different brands, models, years, and vehicle conditions.
Since zero mileage is unusual for used vehicles, these records require further investigation. However, there is not enough evidence to conclude that all of them are incorrect, so the values will be retained for now and considered during data preprocessing.

```python
df[df['mileage_km'] == 0]['condition'].value_counts()
```

🔍🚗 Zero Mileage Conclusion

Among the 190 vehicles with 0 km mileage:

* 🚗 174 have the condition with mileage
* 🔧 14 have the condition for parts
* ⚠️ 2 have the condition with damage

The 174 vehicles listed as with mileage with 0 km are particularly unusual and may indicate missing or incorrectly recorded mileage. However, without reliable evidence of the actual mileage, these values will not be changed at this stage.
